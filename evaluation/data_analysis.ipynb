```python
CLEANED_DATASET_PATH = Path('E:\\tmp\\VKQ\\cls-sync-data-btp\\evaluation\\datasets\\golden_eval.csv')
```

```python
# 1. Strip so_hieu
df['so_hieu'] = df['so_hieu'].str.strip()
```

```python
# 2. Strip reference (leading/trailing + normalize internal spaces)
df['reference'] = df['reference'].str.strip().str.replace(r'  +', ' ', regex=True)
```

```python
df_dedup = df.drop_duplicates(keep="first").reset_index(drop=True)
```

```python
# 4. Review near-duplicates manually before dropping
near_dups = df[df.duplicated(subset=['so_hieu','reference','relation'], keep=False)]
print(f"Number of near-duplicates: {len(near_dups)}")
```

```python
df_dedup = df_dedup.drop_duplicates(subset=['so_hieu','reference','relation'], keep='first')
```

```python
df_dedup.to_csv(CLEANED_DATASET_PATH, index=False)
```

```python
cleaned_df = pd.read_csv(CLEANED_DATASET_PATH)  
print(f"Columns in cleaned_df: {list(cleaned_df.columns)}")
cleaned_df.info()
print(f"Number of duplicated rows in cleaned_df: {cleaned_df.duplicated().sum()}")
```

```python
# 4. Review near-duplicates manually before dropping
near_dups = cleaned_df[cleaned_df.duplicated(subset=['so_hieu','reference','relation'], keep=False)]
print(f"Number of near-duplicates: {len(near_dups)}")
```

```python
# Unique clauses by (so_hieu, clause_type, content, parent_content, grandparent_content)
unique_clauses = cleaned_df.drop_duplicates(subset=['so_hieu', 'clause_type', 'content', 'parent_content', 'grandparent_content'])
print(f"Number of unique clauses: {len(unique_clauses)}")
```

```python
# Unique GT pairs by (so_hieu, clause_type, reference, relation)
unique_gt_pairs = cleaned_df.drop_duplicates(subset=['so_hieu', 'clause_type', 'reference', 'relation'])
print(f"Number of unique GT pairs: {len(unique_gt_pairs)}")
```
